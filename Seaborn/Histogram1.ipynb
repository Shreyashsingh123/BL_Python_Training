```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Histogram plot

```python
data=sns.load_dataset("penguins")
data
```

```python
sns.histplot(data["bill_length_mm"])
plt.show()
```

```python
sns.displot(x="flipper_length_mm", data=data)
plt.show()
```

```python
sns.displot(x="flipper_length_mm", data=data,bins=[170,180,190,200,210,220,230,240])
plt.show()
```

```python
sns.displot(x="flipper_length_mm", data=data,bins=25)
plt.show()
```

```python
# add a kernel density curve on the basis of bar kde
# smooth line showing distribution
sns.displot(x="flipper_length_mm", data=data,kde=True)
plt.show()
```

```python
sns.histplot(data=data, x="flipper_length_mm", hue="sex", kde=True)
plt.grid()
plt.show()
```
