```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

BARPLOT PRACTICE

```python
data=sns.load_dataset("penguins")
# data.head(10)
```

```python
# data=pd.DataFrame(data)
sns.barplot(x="island",y="bill_length_mm",data=data)
plt.show()
```

```python

sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex")
plt.show()
```

```python
# order of graph
order1=["Biscoe","Dream","Torgersen"]
sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex",order=order1)
plt.show()
# similarly we can set hueparameter 
```

```python
# interval set length set of interval
sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex",ci=90)
plt.show()
```

```python
# horizontally representation both should be numerical parameter
sns.barplot(x="flipper_length_mm",y="bill_length_mm",data=data,hue="sex",orient="h")
plt.show()
```

```python
# set color
sns.barplot(x="island",y="bill_length_mm",data=data,color="r")
plt.show()
```

```python
sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex",palette="icefire")
plt.show()
```

```python
# light the color low value light color high value dark color
sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex",palette="icefire",saturation=0.3)
plt.show()
```

```python
# change color of line denoting distribution or width 
sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex",palette="icefire",errcolor="green")
plt.show()
```

```python
sns.barplot(x="island",y="bill_length_mm",data=data,hue="sex",palette="icefire",errcolor="green",capsize=0.5)
plt.show()
```
