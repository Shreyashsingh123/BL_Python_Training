```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

tips = sns.load_dataset("tips")
print(tips.head())
# Create box plot
sns.boxplot(x="day", y="total_bill", data=tips)

plt.show()
```

```python
sns.boxplot(
    x="day",
    y="total_bill",
    data=tips,
    palette="Set2",
    hue="sex"
)

plt.show()
```

```python
# compare distribution across many groups
sns.boxplot(x="day", y="total_bill", hue="time",palette="pastel", data=tips)
plt.show()
```
