```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=[1,2,3,4,5]
y=[10,20,10,40,30]

plt.plot(x,y)
plt.show()
```

```python
sns.lineplot(x=x,y=y)
plt.show()
```

```python
df1=pd.DataFrame({"x":x,"y":y})
sns.lineplot(x="x",y="y",data=df1)
plt.show()
```

```python
data=sns.load_dataset("penguins")
```

```python
data
```

linear plot from above data

```python
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=data,hue="sex")
plt.show()
# hue is to plot diffrent data set from that parameter in x and y 
# like her for male and female diffrent color data chart is plotted
# no need to give xlabel,ylabel and legend
```

```python
# customize size of graph using size parameter
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=data,size=10)
plt.show()
```

```python
# change color of chart using Palette
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=data,hue="sex",palette="Oranges")
plt.show() 
```

```python
# change style of plot
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=data,hue="sex",style="sex")
plt.show()
```

```python
# add marker for male and female both
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=data,hue="sex",style="sex",markers=["o","^"])
plt.show()
```

```python
sns.lineplot(x="bill_length_mm",y="flipper_length_mm",data=data,hue="sex",style="sex",markers=["o","^"])
plt.grid()
plt.show()
```
