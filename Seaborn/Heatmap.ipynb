HEATMAP
It is used to show each cell value as a color density
Shows Rows and columns as grid
Works on 2d data
Cannot make heatmap of categorical value


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=np.linspace(1,10,20).reshape(4,5)
data
```

```python
sns.heatmap(data)
# represent data in form of color
```

```python
df=sns.load_dataset("penguins")
df=df.head(30)
df
```

```python
df2=df.drop(columns=["species","island","sex"]).head(15)
df2
```

```python

sns.heatmap(df2)
```

```python
sns.heatmap(data,vmin=0,vmax=15)
# Set value range of labels after graph
```

```python
# cmap to change color
sns.heatmap(data,cmap="Oranges")
```

```python
# to show value on chart
sns.heatmap(data,cmap="Oranges",annot=True)
# to set your data create list of data use annot =list and fnt=data type of list data
```

```python
# set fonsize nad color of data inside chart 
y={'color':"r","fontsize":25}
sns.heatmap(data,annot=True,annot_kws=y)
```

```python
# gap between lines 
y={'color':"r","fontsize":25}
sns.heatmap(data,annot=True,annot_kws=y,linewidths=10)
```
