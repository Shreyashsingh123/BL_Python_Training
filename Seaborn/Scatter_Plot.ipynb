```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

SCATTER PLOT

```python
data=sns.load_dataset("penguins")
data=data.head(30)
```

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=data,hue="sex")
plt.show()
```

```python
# diffrent size for male and female
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=data,hue="sex",size="sex")
plt.show()
```

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=data,hue="sex",size="sex",sizes=(40,80))
plt.show()
```

```python
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=data,hue="sex",palette="gist_ncar")
plt.show()
```

```python
# blur 
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=data,hue="sex",alpha=0.3)
plt.show()
```

```python
# marker 
m={"Male":"*","Female":"o"}
sns.scatterplot(x="bill_length_mm",y="bill_depth_mm",data=data,hue="sex",palette="gist_ncar",style="sex",markers=m)
# marker always work with style
plt.show()
```
