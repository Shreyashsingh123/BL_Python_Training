```python
ser1=pd.Series([2,4,6,8,10])
ser2=pd.Series([1,3,5,7,9])
sum=ser1+ser2
sub=ser1-ser2
mul=ser1-ser2
div=ser1/ser2
print("Sum of arrays is: ",sum)
print("Subtraction of series is ",sub)
print("Division of series is :",div)
print("Multiplication of series is :",mul)
```
