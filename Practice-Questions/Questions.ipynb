4. Write a Python program to draw line charts of the financial data of Alphabet Inc. between
October 3, 2016 to October 7, 2016.

```python
data=pd.read_csv("fData.csv")
print(data)
# for line in data:
#     v=line.split(",")

x=data['Date']
op=data['Open']
hi=data['High']
lo=data['Low']
cl=data['Close']
plt.plot(x,op)
plt.plot(x,hi)
plt.plot(x,lo)
plt.plot(x,cl)
plt.show()
```

```python
x=[]
y1=[]
y2=[]
y3=[]
y4=[]

with open ("fData.csv",'r') as f:
    next(f) #skip first row
    for line in f:
        v=line.split(",")
        x.append(v[0])
        y1.append(float(v[1]))
        y2.append(float(v[2]))
        y3.append(float(v[3]))
        y4.append(float(v[4]))
plt.plot(x,y1,color='red')
plt.plot(x,y2,color='black')
plt.plot(x,y3,color='blue')
plt.plot(x,y4,color='yellow')
plt.show()
```

5. Write a Python program to plot two or more lines on same plot with suitable legends of each
line.

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
z=[15,25,35,45,55]
z2=[18,28,38,48,56]
plt.plot(x,y,label="X-label")
plt.plot(x,z,label="Z-Label")
plt.plot(x,z2,label='Z2-Label')
plt.xlabel("X Axis")
plt.ylabel("Y axis")
plt.title("Line Graph")
plt.legend()
plt.show()
```

14. Write a Python program to display the grid and draw line charts of the closing value of
Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with
rendering with a larger grid (major grid) and a smaller grid (minor grid).Turn on the grid but turn
off ticks.

```python
data=pd.read_csv("Data2.csv")
x=data['Date']
y=data['Close']
plt.plot(x,y)
plt.grid()
plt.grid(which='minor',ls='-',lw=.8)
plt.minorticks_on()
plt.grid(which='major',ls='-',lw=1.3)

plt.show()
```

15. Write a Python program to create multiple plots.

```python
x=[1,2,3,4,5]
y=[2,3,4,5,6]
z=[1,4,9,12,16]
plt.subplot(2,1,1)# 2 rows one col and first position
plt.plot(x,y)
plt.title("first plot")

plt.subplot(2,1,2) #2 row 1 col 2nd  position
plt.plot(x,z)
plt.title("Second plot")
plt.tight_layout()
plt.show()
```
