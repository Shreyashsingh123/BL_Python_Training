USED PYPLOT API 

```python
x=[1,2,3]
y=[4,5,6]
plt.plot(x,y)#(1,4)(2,5)(3,5)
# plt.show()
plt.grid()#show data in form of grid
plt.show()
```

```python
data={
    'Salary':[0,25000,30000,28000,39000,48000,55000,52000,35000,26000]
}
df=pd.DataFrame(data)
df
```

```python
df.shape#rows ,col in data set
```

LINE PLOT OF ABOVE DATA

```python
plt.plot(df['Salary'])
plt.show()
```

```python
# marker add amarker wherever data is there and colr give that color to graph representation
plt.plot(df['Salary'],color='red',marker='o',linestyle='--',linewidth='2')
# linewidth is to set the width of line linestyle is to set style of line
plt.grid()
plt.show()
```

```python
plt.plot(df['Salary'],marker='o')
f1={'family':'arial','color':'red','size':30}
plt.xlabel("X-Axis",fontdict=f1)
plt.ylabel("Y Axis")
plt.title("Line Chart",fontdict=f1)
plt.show()
```

PLOT USING HISTOGRAM

```python
plt.hist(df['Salary'])
plt.show() #show that how many people are there in a range format 
```

```python
plt.hist(df['Salary'],bins=4,color='green',edgecolor='black')
plt.show() #show that how many people are there in a range format 
```

BOX PLOT

```python
# help to detect outlier in data 
# value far away from shape is outlier
plt.boxplot(df['Salary'])
plt.show()#below part is 255 and red line is 50% and above red line is 75% part from strting of box to ending is Interquartile range
```

Categorical column operations 

```python
df["dept"]=['HR','IT','Sales','Finance','HR']*2
df.head()
```

```python
# use pie 0chart
count=df['dept'].value_counts()
count 
```

```python
plt.pie(count)
plt.show()
```

```python
plt.pie(count,labels=count.index)#pie chart with labels 
plt.show()
```

```python
plt.pie(count,labels=count.index,autopct='%1.1f')#pie chart with percentage 1.1 means 1 decimal 
plt.show()
```

```python
plt.pie(count,labels=count.index,autopct='%1.1f',explode=[0.1,0.1,0.1,0.1])#pie chart separated by 0.1  
plt.show()
```

```python
# to make pie chart a complete circle
plt.pie(count,labels=count.index,autopct='%1.1f',explode=[0.1,0.1,0.1,0.1])#pie chart separated by 0.1  
plt.axis("equal")
plt.show()
```

COUNT PLOT 

```python
plt.bar(count.index,count,color=['green','black','red','blue'])
plt.show() #count by index (HR,Sales etc...)
```
