creating dataframe using series data

```python
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([6,7,8,9,10])
df=pd.DataFrame({'c1':s1,'c2':s2})
# put s1 data in c1 col and s2 data in c2 col as s1 and 2 ar in form of a col 1d
df
```

methods of data frame

```python
print("rows and col is:",df.shape)
print("Column is:",df.columns)
print("Row is:",df.dtypes) #data type of every column 
```

accessing data in data frame

```python
# using labes
df.loc[0,"c1"]
# 0th row and c1 col value
```

```python
print(df.iloc[0,1])
# for index wise access
# oth row and 1st col value
```

 information about data

```python
df.info()
```

```python
df.describe()
```

Filtering data using operators

```python
df[df['c1']>2]
# show data in c1 col greater than 2
```

```python
df[df['c2']<8]
```

add col,update col,delete col and much more operation on dataset

```python
print(df)
df['c3']=[11,12,13,14,15]
df
```

```python
# delete a col 
print(df)
del df['c3']
print(df)
```

```python
# another way to delete
df2=df.drop(columns=['c3'])
print(df2)
```

update value from a column

```python
print(df)
df.loc[df['c1']==2]=9 #update value of 2 to 9 in c1 col
df
```

```python
# rename column name if not used inplace then changes are temporary not in original array
df.rename(columns={'c1':'c0'},inplace=True)
df
```

Set a index

```python
df.set_index('c2',inplace=True)
df #make c2 as index value
```

```python
df.loc[6,'c1'] #6 as index and c3 col 
```

```python
df.iloc[3,0]#3rd index from c2 and index start from 0 to n-1 
```

```python
# reset index
df.reset_index(inplace=True)

```

Multi dimensional data

```python
data1=pd.DataFrame({
    'Park':['A','B','C','B'],
    'Name':["Ram","Shyam","Chrish","Aman"],
    'Age':[18,21,25,28]
}).set_index(['Park','Name'])
data1 #name and age is index here
```

Selecting with condition

```python
data1[data1['Age']>18] #data with age >18
```

alternate wy of selecting with condition

```python
data1.query('Age>18')
```
