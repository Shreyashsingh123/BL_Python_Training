```python
import pandas as pd
import numpy as np
```

Series is used to store 1d data and can be accesed by indexing start from 0

```python
data=pd.Series([1,2,3,4])
print("data is:\n",data)
print("First value is:",data[0])
print("First 3 value is:",data[:3])
```

Information about data Attributes

```python
print(data.index) #give start index nad stop index 
```

```python
print(data.values) #gives values from data
```

```python
print("Datatype is:",data.dtype)
```

Some methods of pandas

```python
print("First 2 data is:",data.head(2))
```

```python
print("Last 2 data is:",data.tail(2))#by default value is 5
```

```python
arr=pd.Series([2,6,1,8,3,9])

print("Sorted data is: ",arr.sort_values())
```

Count total values

```python
print("Total values are:",arr.count())
```

check null values 

```python
na=arr.isnull()
print(na)
print("Count of null value is",arr.isnull().sum())
```

check not null values

```python
arr=pd.Series([2,6,1,8,3,9])
print(arr.notnull())
print("Count of not null is",arr.notnull().count())
```

```python
missing_data=pd.Series([2,None,5])
print(missing_data.isnull())
print(missing_data.notnull())
```

Fill a value in place of null but does not modify original value

```python
new_arr=missing_data.fillna(11)
new_arr
```
